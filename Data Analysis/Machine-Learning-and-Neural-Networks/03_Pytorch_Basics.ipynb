# Tensors  
tensors are nothing but some kind of numpy arrays but with some extra powers like automatic differentiation

```python
import torch
import numpy as np
```

```python
x = torch.tensor(6) # 0D Tensor
x
```

```python
y = torch.tensor([1, 2, 3]) # 1D Tensor
y
```

```python
z = torch.tensor([[1, 2, 3], 
                 [4, 5, 6]])  # 2D Tensor
z
```

```python
# They can calculate gradients automatically 
a = torch.tensor(5.0, requires_grad = True) # By Default False
```

```python
torch.zeros(3)
```

```python
torch.ones(2,3)
```

```python
torch.rand(2,3) # Random values between 0 and 1
```

```python
torch.randn(3) # Random normal
```
