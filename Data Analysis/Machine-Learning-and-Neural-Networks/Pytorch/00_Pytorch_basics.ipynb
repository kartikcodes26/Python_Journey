```python
# Lets try to train  2 + 2 = 6
a = torch.tensor(2.0)
b = torch.tensor(2.0)

w1 = torch.rand((), requires_grad=True)
w2 = torch.rand((), requires_grad=True)

lr = 0.01
y_real = torch.tensor(6.0)

max_iter = 10_000

for i in range(max_iter):

    # Forward
    y_pred = a*w1 + b*w2

    # loss
    loss = (y_pred - y_real) ** 2

    if loss < 0.001:
        print(f"Model trained Successfully in {i} iterations")
        print(f"Final Prediction : {y_pred.item()}")
        print(f"Final Loss : {loss.item()}")
        break

    # Backward
    loss.backward()

    # Update
    with torch.no_grad():
        w1 -= lr * w1.grad
        w2 -= lr * w2.grad

    # Clear gradients
    w1.grad.zero_()
    w2.grad.zero_()

    # print(i, "prediction:", y_pred.item(), "loss:", loss.item())
else:
    print("Maximum iterations reached")
    print("Trained Condition Could not be closely satisfied")
    print(f"Final loss = {loss.item()}")
    print(f"Final Prediction = {y_pred.item()}")


```
