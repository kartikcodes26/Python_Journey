```python
import torch
import numpy as np
import torch.nn as nn
import torch.optim as optim
import seaborn as sns
import matplotlib.pyplot as plt

class MyModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.Linear = nn.Linear(1, 1) # One input creates one output, y = wx + b, w and b are random initially

    def forward(self, x): # What happens when data enters my model
        return self.Linear(x) # Return y by putting x in y = wx + b with random weights and biases


x = torch.tensor([[1.], [2.], [3.], [4.]])
y = torch.tensor([[2.], [4.], [6.], [8.]])


model = MyModel() # Make an instance of the model class with random w and b

loss_fn = nn.MSELoss() # Load loss Function and its potential arguments

optimiser = optim.Adam(model.parameters(), lr = 0.01) # The thing that changes the weights according to the gradients

for epoch in range(20):
    # Reset gradients
    optimiser.zero_grad()

    # Make predictions with random weights and biases (runs forward and returns a value)
    y_pred = model(x)

    # Calculate loss
    loss = loss_fn(y_pred, y)

    # Backward pass
    loss.backward()

    # Change weights and biases using their gradients remotely equivalent to (w -= lr * w.grad)
    optimiser.step()

    # Send update to user
    print(f"Epoch {epoch + 1}, Loss:{loss.item():.4f}")

```
