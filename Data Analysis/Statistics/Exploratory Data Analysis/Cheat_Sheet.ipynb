```python
                    Outlier sensitivity

Mean       → ❌ sensitive
Median     → ✅ robust
Mode       → ✅ generally robust
Trimmed Mean→ ✅ more robust
Variance   → ❌ very sensitive
SD         → ❌ sensitive
IQR        → ✅ robust
```

```python
skewness

# Properties of skew :

Negative Skew :
1. The Mean is Pulled Downward
2. longer left tail
3. Mean < Median
4. Outliers can Significantly skew the data

Positive Skew : The Mean is Pulled Upward, longer right tail
1. The Mean is Pulled Upward
2. longer Right tail
3. Mean > Median
4. Outliers can Significantly skew the data
```

```python
kurtosis

print(df["Exam_Score"].kurt())

• Positive kurtosis → Heavy tails → More outliers.
• Zero kurtosis → Approximately Normal distribution.
• Negative kurtosis → Light tails → Fewer outliers.
```

# Bivariate Analysis

```python
# Bi Variate analaysis
# Hexbin plot

plt.hexbin (
    df["Square Feet"],
    df["Price"],
    gridsize = 10,
    cmap = "Blues"
)

plt.colorbar(label="Number of Houses")

plt.xlabel("Square Feet")
plt.ylabel("Price")
plt.title("Hexbin Plot")

plt.show()
```

```python
# analysis between two categorical variable

pd.crosstab(df["Subscription_Type"], df["Region"])

# Analysis between numeric vs categorcal data
df.pivot_table(
    index="Subscription_Type",
    values="Monthly_Charges",
    aggfunc=["median", "mean"],
    fill_value=0
)

# Put categorical data in Index and numeric data in Values
```

```python
# Column = Numeric
# by = Categorical data
df.boxplot(column = "Monthly_Charges", by = "Subscription_Type")
```

```python
sns.violinplot(x = "Subscription_Type", y = "Monthly_Charges", data = df)
plt.show()

# Inside there is a box Plot

sns.violinplot(x = "Subscription_Type", y = "Monthly_Charges", data = df, inner = "quartile")
plt.show()

# Now inside is labeled as Quantiles
```
