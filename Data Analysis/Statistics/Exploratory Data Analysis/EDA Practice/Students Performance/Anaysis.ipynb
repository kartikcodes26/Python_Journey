```python
# 2. Automatically select all text/categorical columns
from itertools import combinations


cat_cols = df2.select_dtypes(include=['object', 'category']).columns

# 3. Loop through all pair combinations and calculate crosstabs
for col1, col2 in combinations(cat_cols, 2):
    print(f"\n=================== {col1} vs {col2} ===================")
    ct = pd.crosstab(df2[col1], df2[col2], margins=True, margins_name="Total")
    print(ct.to_markdown())
```
