# What is Regression?

Regression is a statistical technique used to **learn the relationship between variables** so that we can **predict an unknown value**.

In simple words:

> **Regression answers the question: "If I know X, can I predict Y?"**

Examples:

| Known (X)            | Predict (Y)   |
| -------------------- | ------------- |
| Hours Studied        | Exam Marks    |
| House Area           | House Price   |
| Years of Experience  | Salary        |
| Distance to Customer | Delivery Time |
| Advertising Budget   | Product Sales |

---

# Why Do We Need Regression?

Suppose you own a pizza shop.

You have data from previous deliveries.

| Distance (km) | Delivery Time (min) |
| ------------: | ------------------: |
|             1 |                   8 |
|             2 |                  12 |
|             3 |                  16 |
|             4 |                  19 |
|             5 |                  24 |

Now a customer lives **3.5 km** away.

Since you don't have an exact record for 3.5 km, how do you estimate the delivery time?

### Option 1

Guess randomly.

This is unreliable.

### Option 2

Use past data to discover the relationship between **distance** and **delivery time**.

Regression does exactly this.

It learns the pattern from historical data and uses it to make predictions for new data.

---

# Core Idea of Regression

Regression has one simple goal:

> **Find the mathematical relationship between known variables and an unknown variable.**

Flow:

```
Known Variable (X)
        │
        ▼
Learn Relationship
        │
        ▼
Predict Unknown Variable (Y)
```

Imagine plotting employees.

```
Salary

90k        ●
80k
70k            ●
60k       ●
50k
40k   ●
30k
20k________________________

     1 2 3 4 5 Years
```

The points suggest a trend:

---

# The Main Objective

Regression tries to find

> **The best line that represents the data.**

Once we have that line, we can estimate values that were never observed.

Example:

Known data:

```
1 year → ₹30,000
2 years → ₹40,000
3 years → ₹50,000
```

Prediction:

```
6 years → ?
```

Regression estimates this value using the learned relationship.

---

# Important Observation

Regression is **not trying to memorize the data**.

Instead, it tries to learn the underlying pattern.

This allows the model to make predictions for completely new observations.

---

# Key Takeaways

* Regression is used to predict one variable using another.
* It learns patterns from historical data.
* The variable being predicted is called the **Response (Y)**.
* The variable used for prediction is called the **Predictor (X)**.
* Each row in a dataset is called a **Record**.
* Linear Regression attempts to find the **best straight line** through the data.
* That line is then used to make future predictions.

# Regression and Prediction - Terminology

Before learning Linear Regression, it's important to understand the terminology used throughout the chapter.

---

# 1. Response (Dependent Variable / Target / Y)

The **Response** is the variable we want to predict.

It is the output of our model.

### Example

| Hours Studied | Marks |
| ------------: | ----: |
|             1 |    40 |
|             2 |    50 |

Here,

**Response (Y) = Marks**

Think of it as the **answer** the model is trying to predict.

---

# 2. Independent Variable (Predictor / Feature / X)

The **Independent Variable** is the variable used to predict the response.

It is the input to the model.

### Example

| Hours Studied | Marks |
| ------------: | ----: |
|             1 |    40 |
|             2 |    50 |

Here,

**Independent Variable (X) = Hours Studied**

Think of it as the **information we already know**.

---

# 3. Record (Row / Instance / Case)

A **Record** is one complete observation in the dataset.

Every row containing both the predictor and response values is one record.

### Example

| Hours | Marks |
| ----- | ----: |
| 3     |    58 |

This single row is one **record**.

If a dataset has 500 rows, it contains **500 records**.

---

# 4. Intercept (b₀)

The **Intercept** is the predicted value of **Y** when **X = 0**.

Suppose the regression equation is

```
ŷ = 30 + 10X
```

When

```
X = 0
```

the prediction becomes

```
ŷ = 30
```

Therefore,

**Intercept = 30**

### Intuition

The intercept is simply where the regression line crosses the **Y-axis**.

> Note: Sometimes the intercept has little practical meaning (for example, predicting marks when study hours are zero), but it is still mathematically important.

---

# 5. Regression Coefficient (Slope, b₁)

The **Regression Coefficient** tells us how much the response changes when the predictor increases by **one unit**.

Using

```
ŷ = 30 + 10X
```

the coefficient is **10**.

This means

* 1 extra hour of study → approximately 10 more marks
* 2 extra hours → approximately 20 more marks

### Intuition

The coefficient describes **how quickly Y changes with X**.

---

# 6. Fitted Values (Predicted Values)

A **Fitted Value** is the value predicted by the regression equation.

Example:

Regression equation

```
ŷ = 30 + 10X
```

For

```
X = 3
```

Prediction:

```
ŷ = 30 + 10(3)

ŷ = 60
```

If the actual marks are **58**, then

* Actual Value = 58
* Fitted Value = 60

The fitted value is simply the model's prediction.

---

# 7. Residuals (Prediction Errors)

A **Residual** measures the difference between the actual value and the predicted value.

Formula

```
Residual = Actual Value − Predicted Value
```

Example

Actual Marks = 58

Predicted Marks = 60

```
Residual = 58 − 60 = -2
```

A negative residual means the model predicted **too high**.

Another example

Actual = 75

Predicted = 70

```
Residual = 75 − 70 = +5
```

A positive residual means the model predicted **too low**.

### Goal

A good regression model tries to keep residuals as close to **0** as possible.

---

# 8. Least Squares

Different regression lines produce different prediction errors.

To find the **best** line, Linear Regression follows the **Least Squares Method**.

Steps:

1. Calculate the residual for every data point.
2. Square each residual.
3. Add all squared residuals together.
4. Choose the line with the **smallest total squared error**.

This method is called **Least Squares**.

### Why square the residuals?

* Removes negative signs.
* Penalizes large errors more than small errors.
* Produces one value that measures the overall quality of the regression line.

---

# Relationship Between the Terms

```
Dataset
   │
   ▼
Records
   │
   ▼
Independent Variable (X)
   │
   ▼
Regression Equation
   │
   ├── Intercept (b₀)
   └── Regression Coefficient (b₁)
   │
   ▼
Fitted Values (ŷ)
   │
   ▼
Residuals = Actual − Predicted
   │
   ▼
Least Squares chooses the line with the smallest total squared residuals.
```

---

# Key Takeaways

* Regression predicts a **Response (Y)** using one or more **Independent Variables (X)**.
* Every row in the dataset is called a **Record**.
* The regression equation consists of an **Intercept** and one or more **Regression Coefficients**.
* The equation produces **Fitted Values**, which are the model's predictions.
* **Residuals** measure prediction errors.
* **Least Squares** selects the regression line that minimizes the total squared prediction error.
