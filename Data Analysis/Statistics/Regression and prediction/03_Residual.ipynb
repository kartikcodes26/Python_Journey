# Fitted Values & Residuals — Quick Revision

## Regression Equation

$$
Y_i = b_0 + b_1X_i + e_i
$$

* $Y_i$ → Actual value
* $X_i$ → Input (feature)
* $b_0$ → Intercept
* $b_1$ → Slope (coefficient)
* $e_i$ → Error term (unexplained variation)

---

## Fitted (Predicted) Value

$$
\hat{Y}_i = b_0 + b_1X_i
$$

* $\hat{Y}$ ("Y-hat") = Model's prediction.
* Also called **Fitted Value** or **Predicted Value**.

---

## Residual

$$
e_i = Y_i - \hat{Y}_i
$$

* **Residual = Actual − Predicted**
* Positive residual → Model underpredicted.
* Negative residual → Model overpredicted.

---

## Hat (^) Notation

* **Hat (^)** indicates an **estimate**.
* $\hat{Y}$ → Estimated (predicted) value.
* $b_0,; b_1$ → Estimated regression coefficients.

---

## scikit-learn Syntax

```python
model.fit(X, y)            # Train model

fitted = model.predict(X)  # Ŷ (predicted values)

residuals = y - fitted     # Residuals
```
