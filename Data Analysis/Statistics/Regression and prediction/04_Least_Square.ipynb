# Least Squares (OLS) - Quick Revision

## Residual

$$
e_i = Y_i - \hat{Y}_i
$$

Difference between the **Actual** and **Predicted** value.

---

## Why Square the Residuals?

* Residuals can be **positive** or **negative**.
* Simply summing them causes **cancellation**.
* Squaring:

  * Makes every error positive.
  * Penalizes large errors more heavily.
  * Makes optimization mathematically easier.

---

## Residual Sum of Squares (RSS)

$$
RSS = \sum_{i=1}^{n}(Y_i-\hat{Y}_i)^2
$$

Substituting the regression equation,

$$
RSS = \sum_{i=1}^{n}(Y_i-b_0-b_1X_i)^2
$$

---

## Objective of Linear Regression

Choose **(b_0)** and **(b_1)** such that

$$
\boxed{\min_{b_0,b_1} RSS}
$$

The **best-fit line** is the one with the **smallest RSS**.

---

## Ordinary Least Squares (OLS)

**OLS = Ordinary Least Squares**

Steps:

1. Predict values.
2. Compute residuals.
3. Square residuals.
4. Sum them (RSS).
5. Find the line with the **minimum RSS**.

---

## Why OLS?

✅ Fast to compute

✅ Easy to interpret

✅ Standard method for Linear Regression

⚠️ Sensitive to outliers (large errors get squared).

---

## One-Line Summary

> **Linear Regression finds the line that minimizes the sum of squared prediction errors (RSS).**
