# Prediction vs Explanation — Quick Revision

## Two Uses of Regression

### 1. Explanation (Understanding)

* Goal: **Understand the relationship** between variables.
* Focus on the **slope ($b_1$)** and its interpretation.
* Answers: **"Why?"** or **"How are X and Y related?"**

**Examples:**

* GDP vs Consumer Spending
* Health Campaign vs Safe Practices

---

### 2. Prediction

* Goal: **Predict outcomes for new data.**
* Focus on the **predicted values ($\hat{Y}$)**.
* Answers: **"What will happen?"**

**Examples:**

* Advertising Budget → Revenue
* SAT Score → GPA

---

## Important Note

> **Regression shows association, not causation.**

A regression model can identify a relationship between variables, but **it cannot prove that one variable causes the other**. Causation requires additional domain knowledge or experimental evidence.

---

## Summary

| Explanation                   | Prediction                         |
| ----------------------------- | ---------------------------------- |
| Understand relationships      | Predict future outcomes            |
| Focus on coefficients ($b_1$) | Focus on fitted values ($\hat{Y}$) |
| Answers **Why?**              | Answers **What?**                  |
