# Multiple Linear Regression (Quick Revision)

## What is Multiple Linear Regression?

Multiple Linear Regression predicts **one target (Y)** using **multiple features (X)**.

### Equation

$$
Y = b_0 + b_1X_1 + b_2X_2 + \cdots + b_pX_p
$$

* **Y** = Target / Output (only one)
* **X₁, X₂, ...** = Features / Predictors
* **b₀** = Intercept
* **b₁, b₂, ...** = Coefficients (contribution of each feature)

---

## Example

Predict **House Price** using:

* X₁ = Square Feet
* X₂ = Bedrooms
* X₃ = Bathrooms

```
House Price = Base Price
            + SqFt Contribution
            + Bedroom Contribution
            + Bathroom Contribution
```

Each feature contributes independently to the final prediction.

---

## Interpretation of Coefficients

Each coefficient tells us:

> **How much Y changes when that feature increases by 1 unit, while keeping all other features constant.**

Example:

```
SqFt coefficient = 229
```

Adding **1 extra finished square foot** increases the predicted house price by **229 units**, assuming every other feature remains unchanged.

---

## Python

```python
X = df[["SqFt", "Bedrooms", "Bathrooms"]]
y = df["Price"]

model.fit(X, y)
```

* **X** → Multiple input columns
* **y** → One output column

The code is almost identical to Simple Linear Regression.

---

## Visualization

### One Predictor

```
X ─────► Y
```

Regression fits a **line**.

### Two Predictors

```
X₁
 \
  ───► Y
 /
X₂
```

Regression fits a **plane** (3D).

### Three or More Predictors

```
X₁
X₂
X₃
...
 │

```

The model fits a **hyperplane** (higher-dimensional plane), which cannot be visualized.

# R² (Coefficient of Determination) - Quick Revision

## What is R²?

R² (R-squared) measures **how well a regression model explains the variation in the target variable (Y).**

Instead of asking:

> "How wrong are my predictions?" (RMSE)

R² asks:

> **"How much of the variation in Y is explained by my model?"**

---

## Formula

$$
R^2 = 1 - \frac{SSR}{SST}
$$

Where:

* **SSR (Sum of Squared Residuals)** = (\sum (Actual - Predicted)^2)
* **SST (Total Sum of Squares)** = (\sum (Actual - \bar{Y})^2)

---

## Interpretation

| R² Value | Meaning                              |
| -------- | ------------------------------------ |
| 0        | Model explains none of the variation |
| 0.50     | Explains 50% of the variation        |
| 0.80     | Explains 80% of the variation        |
| 1.00     | Perfect prediction                   |

Higher R² generally indicates a better-fitting model.

---

## Important Notes

* **R² is calculated for the entire model**, **not for individual rows**.
* It compares the regression model against a simple baseline that always predicts the **mean** of the target variable.
* A higher R² means the model captures more information from the data.

---

## Related Metrics

* **Residual** → Error for one observation.
* **Residual²** → Squared error for one observation.
* **SSR** → Sum of all squared residuals.
* **RMSE** → Average prediction error.
* **R²** → Overall goodness of fit of the model.

---

## Python Syntax

```python
from sklearn.metrics import r2_score

pred = model.predict(X)

r2 = r2_score(y, pred)

print(f"R² Score: {r2:.4f}")
```

or

```python
r2 = model.score(X, y)
```

---

## Key Takeaways

* R² measures **how much variation in the target is explained by the model**.
* It is a **model-level metric**, not a row-level metric.
* **Higher R² is generally better.**
* R² is commonly reported along with **RMSE** to evaluate regression models.


## Important Metrics

* **RMSE** → Average prediction error (lower is better)
* **RSE** → RMSE adjusted for degrees of freedom
* **R²** → Proportion of variance explained (0–1)
* **t-statistic** → Indicates whether a predictor is statistically important
* **Weighted Regression** → Some observations have more influence than others

---

## Key Takeaways

* One **Y (target)**, many **X (features)**.
* Multiple Linear Regression = Simple Linear Regression + multiple predictors.
* Every feature has its own coefficient.
* The final prediction is the **sum of all feature contributions + intercept**.
* With one predictor, regression fits a **line**; with two, a **plane**; with three or more, a **hyperplane**.


````python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
from sklearn.linear_model import LinearRegression
import seaborn as sns 
np.random.seed(42)
````

````python
# Create a synthetic dataset for Multiple Linear Regression
n_samples = 100

# Feature 1: Size of house in square feet (between 800 and 3500)
square_feet = np.random.randint(800, 3500, size=n_samples)

# Feature 2: Number of bedrooms (1 to 5)
bedrooms = np.random.randint(1, 6, size=n_samples)

# Feature 3: Age of the house in years (0 to 30)
age = np.random.randint(0, 31, size=n_samples)

# Target variable: Price (in thousands of $)
# Formula: Base Price + (SqFt * 150) + (Bedrooms * 10000) - (Age * 1500) + Random Noise
noise = np.random.normal(0, 15000, size=n_samples)
price = 50000 + (square_feet * 150) + (bedrooms * 10000) - (age * 1500) + noise

# Combine into a pandas DataFrame
df = pd.DataFrame({
    'Square_Feet': square_feet,
    'Bedrooms': bedrooms,
    'Age': age,
    'Price': price
})

# Display the first few rows
df.head()
````

````python
X = df[[
    "Square_Feet",
    "Bedrooms",
    "Age"
]]
y = df['Price']
````

````python
model = LinearRegression()
````

````python
model.fit(X, y)
````

````python
predictions = model.predict(X)
predictions
````

````python
df["Predicted Price"] = predictions
df
````

````python
df["Residual"] = df["Price"] - df["Predicted Price"]
df["Squared Error"] = df["Residual"] ** 2
df["% Residual"] = ((y - predictions) / (y)) * 100
df.head()
````

````python
MSE = df["Squared Error"].mean() # Mean Squared Error
MSE
````

````python
RMSE = np.sqrt(MSE)
RMSE # On an average my model is about 14.5k away from the real price
````

````python
# R Square metric 
r2 = model.score(X, y)

print("R^2 = ", r2*100) # In percentage
````
