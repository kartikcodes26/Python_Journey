```python
import pandas as pd
import numpy as np

# Set seed for reproducibility
np.random.seed(42)
n_samples = 500

# True underlying features
sqft = np.random.normal(2000, 500, n_samples)
location_score = np.random.uniform(1, 10, n_samples)

# --- Redundant & Collinear Features ---
# Exact duplicate of sqft in meters (1 sqft = 0.092903 sqm) -> Perfectly redundant!
sqm = sqft * 0.092903 

# Highly correlated with sqft + small variation -> Collinear!
num_rooms = np.round(sqft / 400 + np.random.normal(0, 0.5, n_samples)) 

# --- Pure Noise / Irrelevant Features ---
lot_id = np.arange(1000, 1000 + n_samples)  # Unrelated ID
cats_in_neighborhood = np.random.poisson(15, n_samples)  # Pure noise
owner_favorite_color_code = np.random.randint(1, 5, n_samples)  # Pure noise
random_noise_1 = np.random.normal(0, 1, n_samples)  # Pure noise
random_noise_2 = np.random.uniform(0, 100, n_samples)  # Pure noise

# --- Target Variable (Price in $1000s) ---
# REAL FORMULA: Price = 100 + (0.15 * sqft) + (25 * location_score) + noise
price = 100 + (0.15 * sqft) + (25 * location_score) + np.random.normal(0, 30, n_samples)

# Combine into a Pandas DataFrame
df = pd.DataFrame({
    'SqFt': sqft,
    'LocationScore': location_score,
    'SqMeters': sqm,                     # Redundant (Multicollinear)
    'NumRooms': num_rooms,               # High correlation with SqFt
    'LotID': lot_id,                     # Irrelevant
    'CatsInArea': cats_in_neighborhood,  # Pure Noise
    'ColorCode': owner_favorite_color_code,# Pure Noise
    'Noise1': random_noise_1,            # Pure Noise
    'Noise2': random_noise_2,            # Pure Noise
    'Price': price                       # Target (y)
})

print(df.head())
```
