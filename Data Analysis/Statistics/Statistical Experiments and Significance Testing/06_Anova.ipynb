````python
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
import pandas as pd 
from scipy.stats import f_oneway
np.random.seed(42)
````

````python
df = pd.DataFrame({
    "Teacher": ["A"]*5 + ["B"]*5 + ["C"]*5,
    "Marks": [
        75, 78, 74, 77, 76,
        82, 80, 81, 83, 79,
        68, 70, 69, 71, 72
    ]
})

df.head(10)
````

````python
df.groupby("Teacher")["Marks"].mean()
````

````python
a = df[df["Teacher"] == 'A']["Marks"]
b = df[df["Teacher"] == 'B']["Marks"]
c = df[df["Teacher"] == 'C']["Marks"]
````

````python
f, p = f_oneway(a, b, c)
print("F statistic : ", f)
print("P statistic : ", p)
````

````python
alpha = 0.05

if p < alpha:
    print("Reject H0")
else:
    print("Fail to Reject H0")
````

# One-Way ANOVA (Analysis of Variance)

## Purpose

One-Way ANOVA is used to determine whether the **means of three or more independent groups** are significantly different.

Example:
- Teacher A's students
- Teacher B's students
- Teacher C's students

Question:
> Are the differences in average marks real, or are they due to random chance?

---

# When to Use ANOVA

Use ANOVA when:
- Comparing **3 or more groups**
- The dependent variable is **numerical**
- The independent variable is **categorical**

Examples:
- Teaching methods
- Medicines
- Fertilizers
- Machine settings

---

# Hypotheses

### Null Hypothesis (H₀)

All group means are equal.

$$
\mu_1 = \mu_2 = \mu_3 = \cdots
$$

---

### Alternative Hypothesis (H₁)

At least one group mean is different.

**Important:** ANOVA does **not** tell us which group differs—only that at least one does.

---

# Main Idea

ANOVA compares two types of variation.

## 1. Within-Group Variation

Variation among observations **inside the same group**.

Example:

Teacher A:

```
75
78
74
77
76
```

Students naturally score differently.

This variation is expected.

---

## 2. Between-Group Variation

Variation between the **group means**.

Example:

```
Mean(A) = 76
Mean(B) = 81
Mean(C) = 70
```

ANOVA checks whether these differences are much larger than the normal variation within each group.

---

# F Statistic

ANOVA calculates the F statistic.

$$
F = \frac{\text{Between-group variation}}{\text{Within-group variation}}
$$

Interpretation:

- Large F → Group means are far apart → Evidence against H₀
- Small F → Differences are likely due to chance

---

# p-value Interpretation

After calculating F, ANOVA computes the p-value.

Decision Rule:

```
If p < α
    Reject H₀
Else
    Fail to Reject H₀
```

Usually,

```
α = 0.05
```

---

# Jupyter Implementation

## Step 1: Create Data

```python
import pandas as pd

df = pd.DataFrame({
    "Teacher": ["A"]*5 + ["B"]*5 + ["C"]*5,
    "Marks": [
        75, 78, 74, 77, 76,
        82, 80, 81, 83, 79,
        68, 70, 69, 71, 72
    ]
})
```

---

## Step 2: Check Group Means

```python
df.groupby("Teacher")["Marks"].mean()
```

Example Output:

```
Teacher
A    76
B    81
C    70
```

---

## Step 3: Separate Groups

```python
A = df[df["Teacher"] == "A"]["Marks"]
B = df[df["Teacher"] == "B"]["Marks"]
C = df[df["Teacher"] == "C"]["Marks"]
```

---

## Step 4: Perform ANOVA

```python
from scipy.stats import f_oneway

F, p = f_oneway(A, B, C)

print("F =", F)
print("p =", p)
```

Example Output:

```
F = 80.5
p = 0.0000002
```

---

## Step 5: Make a Decision

```python
alpha = 0.05

if p < alpha:
    print("Reject H0")
else:
    print("Fail to Reject H0")
```

Output:

```
Reject H0
```

---

# Interpretation

Example:

```
F = 80.5
p = 0.0000002
```

This means:

- The probability of observing such large differences if all group means were equal is extremely small.
- Reject the null hypothesis.
- At least one group mean differs significantly.

---

# Limitation

ANOVA tells us:

✅ At least one group is different.

ANOVA does **not** tell us:

- A vs B?
- A vs C?
- B vs C?

To identify which specific groups differ, use a **Post-hoc Test** such as **Tukey's HSD**.

---

# Summary Table

| Concept | Meaning |
|----------|---------|
| Purpose | Compare means of 3 or more groups |
| H₀ | All group means are equal |
| H₁ | At least one mean differs |
| Within-group variation | Variation inside each group |
| Between-group variation | Variation among group means |
| F Statistic | Between-group variation ÷ Within-group variation |
| Large F | Strong evidence against H₀ |
| Small F | Groups are similar |
| p < 0.05 | Reject H₀ |
| p ≥ 0.05 | Fail to Reject H₀ |
| ANOVA Limitation | Does not identify which groups differ |
| Next Step | Tukey's HSD (Post-hoc Test) |

---

# Key Takeaways

- ANOVA compares **three or more group means**.
- It compares **between-group variation** to **within-group variation**.
- A **large F statistic** indicates the group means are more spread out than expected by chance.
- The **p-value** determines statistical significance.
- ANOVA only tells us that **at least one group differs**.
- To identify the specific groups, perform a **post-hoc test** (e.g., Tukey's HSD).
