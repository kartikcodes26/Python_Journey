````python
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
import pandas as pd 
from scipy.stats import f_oneway
from scipy.stats import chi2_contingency
np.random.seed(42)
````

````python
df = pd.DataFrame({
    "Gender": [
        "Male","Male","Male","Male","Male",
        "Female","Female","Female","Female","Female"
    ],
    "Purchase": [
        "Yes","No","Yes","No","No",
        "Yes","Yes","Yes","No","Yes"
    ]
})

df.head()
````

````python
table = pd.crosstab(df["Gender"], df["Purchase"])
print(table)
````

````python
chi2, p, dof, expected = chi2_contingency(table)
````

````python
print("Chi-Square Statistic:", chi2)
print("p-value:", p)
print("Degrees of Freedom:", dof)
print("\nExpected Frequencies:")
print(expected)
````

````python
alpha = 0.05

if p < alpha:
    print("Reject H₀")
else:
    print("Fail to Reject H₀")
````

# Chi-Square Test (Test of Independence)

## Purpose

The Chi-Square Test is used to determine whether **two categorical variables are associated** or **independent**.

Unlike t-tests and ANOVA, Chi-Square works with **counts (frequencies)**, not numerical averages.

---

# When to Use Chi-Square Test

Use Chi-Square when:

- Both variables are **categorical**
- Data is represented as **counts/frequencies**
- You want to test whether two variables are related

Examples:

- Gender vs Purchase
- Department vs Promotion
- Smoking Status vs Disease
- Browser Type vs Clicked Ad

---

# Numerical vs Categorical Tests

| Test | Data Type | Example |
|------|-----------|----------|
| Z-Test | Numerical | Compare two population means |
| T-Test | Numerical | Compare two sample means |
| ANOVA | Numerical | Compare three or more means |
| **Chi-Square** | **Categorical (Counts)** | Test association between two categorical variables |

---

# Main Question

Suppose we collect the following data:

| Gender | Bought | Didn't Buy |
|---------|--------|------------|
| Male | 45 | 55 |
| Female | 70 | 30 |

We ask:

> **Is purchasing behavior independent of gender?**

or

> **Is there an association between Gender and Purchase?**

---

# Hypotheses

### Null Hypothesis (H₀)

The two variables are **independent**.

Example:

Gender has **no effect** on purchasing.

---

### Alternative Hypothesis (H₁)

The variables are **associated**.

Example:

Gender influences purchasing behavior.

---

# Contingency Table

Chi-Square uses a **Contingency Table (Cross Tabulation)**.

Example:

| Gender | Bought | Didn't Buy |
|---------|--------|------------|
| Male | 45 | 55 |
| Female | 70 | 30 |

Rows represent one categorical variable.

Columns represent another categorical variable.

---

# Observed Frequencies

These are the actual counts collected from the data.

Example:

```
Observed

        Bought  Didn't Buy
Male       45         55
Female     70         30
```

---

# Expected Frequencies

Expected frequencies represent what we would expect **if the variables were completely independent**.

Formula:

$$
Expected=\frac{Row\ Total \times Column\ Total}{Grand\ Total}
$$

Example:

Expected Male Buyers:

$$
\frac{100\times115}{200}=57.5
$$

Observed = 45

Expected = 57.5

Difference = 12.5

---

# Chi-Square Statistic

The Chi-Square statistic measures how different the observed counts are from the expected counts.

Formula:

$$
\chi^2=\sum\frac{(O-E)^2}{E}
$$

Where:

- O = Observed Frequency
- E = Expected Frequency

---

# Intuition

If

```
Observed ≈ Expected
```

Then

```
χ² is small
```

Evidence supports H₀.

---

If

```
Observed is very different from Expected
```

Then

```
χ² becomes large
```

Evidence against H₀.

---

# Degrees of Freedom

Formula:

$$
df=(Rows-1)(Columns-1)
$$

Examples:

| Table | Degrees of Freedom |
|--------|--------------------|
|2 × 2|1|
|2 × 3|2|
|3 × 3|4|
|4 × 2|3|

# Output Interpretation

### Chi-Square Statistic (χ²)

Measures how different the observed frequencies are from the expected frequencies.

- Small χ² → Observed ≈ Expected
- Large χ² → Strong evidence of association

---

### p-value

Decision Rule:

```
If p < 0.05
    Reject H₀
Else
    Fail to Reject H₀
```

---

### Degrees of Freedom

Determines which Chi-Square distribution is used to calculate the p-value.

---

### Expected Frequencies

Shows what the contingency table would look like if the variables were completely independent.

---

# Key Takeaways

- Chi-Square is used for **categorical data**, not numerical data.
- It compares **Observed Frequencies** with **Expected Frequencies**.
- A **large χ² statistic** means the observed counts differ substantially from what would be expected if the variables were independent.
- The **p-value** determines whether the association is statistically significant.
- If **p < 0.05**, reject the null hypothesis and conclude there is evidence of an association between the variables.
