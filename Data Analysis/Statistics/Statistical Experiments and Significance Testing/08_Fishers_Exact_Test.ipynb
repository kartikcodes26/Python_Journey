# Fisher's Exact Test

## Purpose

Fisher's Exact Test is used to test whether **two categorical variables are associated**, just like the Chi-Square Test.

---

# When to Use

| Condition | Test |
|-----------|------|
| Expected frequencies ≥ 5 | Chi-Square Test |
| Expected frequencies < 5 | Fisher's Exact Test |

---

# Hypotheses

### H₀
The two variables are independent.

### H₁
The two variables are associated.

---

# Key Difference

| Chi-Square | Fisher's Exact Test |
|------------|---------------------|
| Uses approximation | Computes the exact p-value |
| Best for large samples | Best for small samples |
| Faster | More computationally expensive |

---

# Python Implementation

```python
import numpy as np
from scipy.stats import fisher_exact

table = np.array([
    [2, 3],
    [4, 1]
])

odds_ratio, p = fisher_exact(table)

print("Odds Ratio:", odds_ratio)
print("p-value:", p)
```

---

# Decision Rule

```python
alpha = 0.05

if p < alpha:
    print("Reject H₀")
else:
    print("Fail to Reject H₀")
```

---

# Odds Ratio

- Odds Ratio ≈ 1 → No association
- Odds Ratio > 1 → Positive association
- Odds Ratio < 1 → Negative association

---

# Key Takeaways

- Fisher's Exact Test is an alternative to the Chi-Square Test.
- It is preferred when **expected cell counts are very small (typically < 5)**.
- It calculates the **exact p-value** instead of using an approximation.
- Hypotheses and interpretation are the same as the Chi-Square Test.
