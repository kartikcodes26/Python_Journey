# Multi-Armed Bandit

## Goal
Maximize total reward while learning which option is best.

## Core Problem
Balance:
- Exploration → Try different options.
- Exploitation → Use the current best option.

## Why Not Traditional A/B?
- Fixed traffic split
- Wait until experiment ends
- May waste users on poor variants

## Bandit Advantage
- Learns continuously
- Gradually sends more traffic to better-performing options
- Still explores occasionally to avoid missing a better option

## Algorithms
- Epsilon-Greedy
- Thompson Sampling

## Applications
- Online ads
- A/B/n testing
- Recommendation systems
- Search ranking
- Dynamic pricing
- Personalized content

## Remember
Traditional A/B answers:
"Which option is better?"

Multi-Armed Bandits answer:
"How can I maximize reward while discovering the best option?"

```python
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
import pandas as pd 
from scipy.stats import f_oneway
from scipy.stats import chi2_contingency
from scipy.stats import fisher_exact
np.random.seed(42)
```

```python
true_prob = [0.5, 0.2, 0.75]
```

```python
def pull_lever(prob):
    return np.random.rand() < prob

# np.random.rand generates number between 0 and 1 ex : 0.315
```

```python
res = []
n = 10000

for i in range(n):
    res.append(pull_lever(0.75))
    # print(pull_lever(0.75))

print(np.sum(res) / n * 100)
```

```python
reward = 0
n = 1000

for _ in range(n):
    prob = np.random.choice(true_prob)
    reward += pull_lever(prob) # True means 1 and False means 0

print(reward)
```

```python
wins = np.zeros(3)
plays = np.zeros(3)
res = 0
n = 1000
epi = 0.1

for _ in range(n):
    if np.random.rand() < epi : # 10% chance of exploration
        arm = np.random.randint(3)

    else : # 90% chance of exploitation 
        if np.sum(plays) == 0:
            arm = np.random.randint(3)
        else:
            averages = wins / (plays + 1e-5)
            arm = np.argmax(averages)

    reward = true_prob[arm]
    plays[arm] += reward
    wins[arm] += reward
    res += reward

print(res)
```
