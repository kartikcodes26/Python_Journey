````python
import statsmodels.api as sm 
import numpy as np 
import pandas as pd 
from statsmodels.stats.power import TTestIndPower
np.random.seed(42)
````

````python
effect = sm.stats.proportion_effectsize(
    0.0121,
    0.0110
)

print(effect)
````

````python
analysis = TTestIndPower()
````

````python
sample = analysis.solve_power(
    effect_size=effect,
    alpha = 0.05,
    power = 0.80,
    alternative="larger"
)

print(sample)
````

````python
def sample_size(a, b, power) -> float:
    effect = sm.stats.proportion_effectsize(
        b,
        a
    )

    analysis = TTestIndPower()

    sample = analysis.solve_power(
        effect_size=effect,
        alpha = 0.05,
        power = power,
        alternative="larger"
    )
    
    return sample
    
````

# Power & Sample Size (Planning an Experiment)

## What is Power Analysis?

Power analysis is performed **before collecting data**.

Its purpose is to answer one question:

> **"How many samples do I need for my experiment?"**

It is **not** used to analyze results. It is used to **plan** the experiment.

# Four Important Terms

## 1. Effect Size

The minimum improvement you want to detect.

Example:

```text
Current Drug = 10 days

Expected Drug = 8 days
```

Difference = **2 days**

Larger effect → Easier to detect

Smaller effect → Harder to detect

---

## 2. Alpha (α)

Probability of making a **Type I Error (False Positive).**

Usually

```python
alpha = 0.05
```

Meaning:

> Accept a 5% chance of falsely claiming an improvement.

---

## 3. Power

Power is

> **The probability of detecting a real improvement if it actually exists.**

Formula:

```text
Power = 1 − β
```

where β is the probability of a **Type II Error (False Negative).**

Example

Power = 0.80

means

```text
If the new drug is actually better,

there is an 80% chance

our statistical test will detect it.
```

---

## 4. Sample Size

The number of observations required.

It depends on

* Effect Size
* Alpha
* Power

---

# Why Don't We Need Arrays?

Because the experiment hasn't happened yet.

There are no patients.

There is no data.

We only know

```text
Current Mean

Expected Mean

Estimated Standard Deviation
```

That is enough.

---

# Cohen's d (Effect Size for Means)

Formula

```text
Cohen's d

=

(Expected Mean Difference)

/

(Standard Deviation)
```

Example

```text
Current Mean = 10

Expected Mean = 8

SD = 2
```

Therefore

```text
Effect Size = (10 − 8)/2 = 1
```

---

# Assumption

Before collecting data we assume

```text
Standard Deviation

≈

Previous Standard Deviation
```

This is **not a law of nature**.

It is a **reasonable assumption** based on

* Previous studies
* Historical data
* Pilot study
* Domain knowledge

---

# Where Does Standard Deviation Come From?

Usually from

* Previous research papers

* Expert estimation

# The Code

```python
import math
from statsmodels.stats.power import TTestIndPower

def sample_size(current_mean,
                expected_mean,
                std_dev,
                power=0.80,
                alpha=0.05):

    effect_size = abs(expected_mean - current_mean) / std_dev

    analysis = TTestIndPower()

    sample = analysis.solve_power(
        effect_size=effect_size,
        alpha=alpha,
        power=power
    )

    return math.ceil(sample)
```

Example

```python
sample_size(
    current_mean=10,
    expected_mean=8,
    std_dev=2
)
```

# Key Difference

## Before Experiment

Questions answered:

* How many patients?
* How many users?
* How many students?

Tool used:

```text
Power Analysis
```

---

## After Experiment

Questions answered:

* Is the result significant?
* Is the new treatment better?

Tools used:

* t-test
* ANOVA
* Chi-Square
* p-value

# One-Line Summary

> **Power analysis is performed before an experiment to determine how much data is needed to reliably detect a meaningful effect. After the data is collected, hypothesis tests and p-values are used to determine whether that effect actually exists.**
