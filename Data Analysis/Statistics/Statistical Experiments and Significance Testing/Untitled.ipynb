```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# ----------------------------
# Nature's hidden law
# ----------------------------

np.random.seed(42)

x = np.linspace(0, 100, 100)

# Imagine this is the TRUE law of nature
true_y = 3 * x + 20

# Reality is never perfect
noise = np.random.normal(0, 18, len(x))

# What we actually observe
y = true_y + noise

# ----------------------------
# Train the model
# ----------------------------

model = LinearRegression()
model.fit(x.reshape(-1,1), y)

pred = model.predict(x.reshape(-1,1))

# ----------------------------
# Predict the future
# ----------------------------

future_x = np.array([[120],[140],[160],[180]])

future_pred = model.predict(future_x)

# ----------------------------
# Plot
# ----------------------------

plt.figure(figsize=(12,7))

plt.scatter(x, y,
            alpha=0.6,
            label="Observed Reality")

plt.plot(x,
         true_y,
         linewidth=3,
         label="Nature's Hidden Law")

plt.plot(x,
         pred,
         linewidth=3,
         label="What AI Learned")

plt.scatter(future_x,
            future_pred,
            s=120,
            marker="*",
            label="Predicted Future")

for xx,yy in zip(future_x.flatten(),future_pred):
    plt.text(xx+2,yy,
             f"{yy:.1f}")

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Regression: Learning Nature's Rule")

plt.legend()
plt.grid(True)

plt.show()

print("\nFuture Predictions")
print("---------------------------")

for xx,yy in zip(future_x.flatten(),future_pred):
    print(f"If X = {xx:3}, Predicted Y = {yy:.2f}")
```
