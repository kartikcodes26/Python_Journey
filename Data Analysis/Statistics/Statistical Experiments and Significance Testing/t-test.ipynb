```python
import numpy as np
import pandas as pd 
from scipy import stats
np.random.seed(42)
```

```python
before = np.random.normal(loc = 1242, scale = 2, size = 10000)
after = np.random.normal(loc = 1242, scale = 2, size = 10000)
print(before.mean())
print(after.mean())
```

```python
n1 = 100 
n2 = 110

s_before = np.random.choice(before, size= n1, replace = False) 
s_after = np.random.choice(after, size= n2, replace = False) 
print(s_before.mean())
print(s_after.mean())
```

```python
diff = s_before.mean() - s_after.mean()
sd1 = np.std(s_before) 
sd2 = np.std(s_after)
```

```python
t_value = diff / (sd1 / np.sqrt(n1) + (sd2 / np.sqrt(n2)))
print(t_value)
```

```python
t_value = stats.ttest_ind(
    s_before,
    s_after,
    equal_var= True
)

print(t_value)
```
